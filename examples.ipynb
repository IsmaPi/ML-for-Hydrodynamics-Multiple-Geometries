```python
import libMobility as lm
import numpy as np

a = eta = 1.0
# solver = lm.SelfMobility(periodicityX="open", periodicityY="open", periodicityZ="open")
solver = lm.NBody(periodicityX="open", periodicityY="open", periodicityZ="open")
solver.initialize(viscosity=eta, hydrodynamicRadius=a)

distances = np.linspace(0, 10, 100)
```

```python
def mobility_truth(r, solver):

    positions = np.array([[0,0,0],[r,0,0]],dtype=np.float32)
    forces = np.array([[1,0,0],[0,0,0]],dtype=np.float32)

    solver.setPositions(positions)
    mf, _ = solver.Mdot(forces=forces)

    return mf[1, :]
```

```python
s = [mobility_truth(r, solver)[0] for r in distances]
```

```python
import matplotlib.pyplot as plt

plt.plot(distances, s)
plt.xlabel("Distance")
plt.ylabel("Mobility")
plt.legend(['x', 'y', 'z'])
plt.show()
```
